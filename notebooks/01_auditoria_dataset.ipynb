```python
def detectar_outliers_iqr(df, columnas):
    resultados = []

    for columna in columnas:
        Q1 = df[columna].quantile(0.25)
        Q3 = df[columna].quantile(0.75)
        IQR = Q3 - Q1

        limite_inferior = Q1 - 1.5 * IQR
        limite_superior = Q3 + 1.5 * IQR

        outliers = df[
            (df[columna] < limite_inferior) |
            (df[columna] > limite_superior)
        ][columna].count()

        resultados.append({
            "Variable": columna,
            "Q1": Q1,
            "Q3": Q3,
            "IQR": IQR,
            "Límite inferior": limite_inferior,
            "Límite superior": limite_superior,
            "Outliers": outliers,
            "% Outliers": (outliers / len(df)) * 100
        })

    return pd.DataFrame(resultados).sort_values(
        "% Outliers",
        ascending=False
    )
```

```python
columnas_numericas = df.select_dtypes(include=np.number).columns.tolist()

# No analizamos el ID como variable clínica
columnas_numericas.remove("id")

outliers = detectar_outliers_iqr(df, columnas_numericas)

outliers
```

```python
def mostrar_outliers(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    resultado = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ].copy()

    return resultado.sort_values(columna)


outliers_glyhb = mostrar_outliers(df, "glyhb")

outliers_glyhb[
    ["id", "glyhb", "stab.glu", "age", "gender", "weight", "waist", "hip"]
]
```

```python
outliers_glucosa = mostrar_outliers(df, "stab.glu")

outliers_glucosa[
    ["id", "stab.glu", "glyhb", "age", "gender", "weight", "waist", "hip"]
]
```

```python
# Revisar posibles valores fuera de rangos básicos

reglas = {
    "age": (0, 120),
    "height": (0, 100),
    "weight": (0, 500),
    "chol": (0, 1000),
    "hdl": (0, 300),
    "stab.glu": (0, 1000),
    "glyhb": (0, 30),
    "bp.1s": (0, 400),
    "bp.1d": (0, 300),
    "bp.2s": (0, 400),
    "bp.2d": (0, 300),
    "waist": (0, 200),
    "hip": (0, 200),
    "time.ppn": (0, 5000)
}

for variable, (minimo, maximo) in reglas.items():
    valores_invalidos = df[
        (df[variable] < minimo) |
        (df[variable] > maximo)
    ][["id", variable]]

    print(f"\n===== {variable} =====")
    print("Valores fuera del rango:", len(valores_invalidos))
    display(valores_invalidos.head(10))
```

```python
# ============================================================
# COMPROBACIÓN DE VARIABLE OBJETIVO
# ============================================================

posibles_targets = [
    "target",
    "diabetes",
    "outcome",
    "diagnosis",
    "diagnostico",
    "type",
    "tipo",
    "diabetes_type",
    "class"
]

targets_encontrados = [
    columna for columna in posibles_targets
    if columna in df.columns
]

print("Columnas del dataset:")
print(df.columns.tolist())

print("\nPosibles variables objetivo encontradas:")
print(targets_encontrados)

if len(targets_encontrados) == 0:
    print("\n⚠️ NO se encontró una variable objetivo.")
    print("Este dataset no está listo para clasificación supervisada.")
else:
    print("\nVariables objetivo encontradas:")
    print(targets_encontrados)
```
